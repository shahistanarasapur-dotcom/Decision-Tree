```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("/content/tennis_anyone_1000.xlsx")
```

```python
df.head()
```

```python
df.columns
```

```python
df.drop(columns = "Day ID",inplace=True)
```

```python
df.nunique()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['Outlook']=le.fit_transform(df['Outlook']) #overcast -0,rain -1,sunny -2
df['Temperature']=le.fit_transform(df['Temperature']) #cool -0,mild-2,hot -1`
df['Humidity']=le.fit_transform(df['Humidity']) #normal -1,high =0
df['Wind']=le.fit_transform(df['Wind']) #strong -0,weak -1
df['Play Tennis']=le.fit_transform(df['Play Tennis']) #yes =1, no -1
```

```python
df.head()
```

```python
### renaming the columns
df.rename(columns={'Play Tennis':'PlayTennis'},inplace =True)
```

```python
df['PlayTennis'].value_counts()
```

```python
df.shape
```

```python
###### splitting dataset vertically (input /output)
X = df.iloc[:,0:-1]
Y = df.iloc[:,-1]
```

```python
###### splitting dataset horizontally(train and test set(20%))
from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=1)
```

```python
#### using decision Tree model
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier()
dt
```

```python
#### training the model
dt.fit(X_train, Y_train)
```

```python
#### predict

predict = dt.predict(X_test)
```

```python
### evaluating model
from  sklearn.metrics import classification_report

print(classification_report(Y_test,predict))
```

>>> Decision Tree accuracy with 100%
